```python
def make_visual(times_local, object_names, object_alts, chosen_objs,legend_out=False):

    fig, ax = plt.subplots(figsize=(6, 4))

    for i in range(len(object_alts)):
        alts = object_alts[i]
    
        if object_names[i] in chosen_objs:   
            ax.plot(times_local, alts, label=object_names[i],lw=5,alpha=1)
        else:
            ax.plot(times_local, alts, label=object_names[i],lw=2.5,alpha=0.5)
            
    ax.axhline(30, linestyle="--")

    if legend_out:
        ax.legend(
            frameon=False,
            loc="lower center",       # anchor the legend relative to bbox
            bbox_to_anchor=(0.5, 1.06),  # x=0.5 centers it, y>1 moves it above the plot
            ncol=4                     # number of columns
        )
    else:
        ax.legend(frameon=False,ncol=4,loc="lower left")

    ax.set_ylabel("Altitude (deg)")
    ax.set_xlabel("Local Time (PST)")
    ax.set_ylim(0, 90)

    # Format x-axis to HH:MM only
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M"))
    ax.xaxis.set_major_locator(mdates.HourLocator(interval=1))

    plt.xticks(rotation=30)

    # Add date range to title
    start_date = times_local[0].strftime("%d %b")
    end_date   = times_local[-1].strftime("%d %b %Y")

    ax.set_title(f"Objects Altitude – Palo Alto ({start_date}–{end_date})")


    plt.show()
    
    return
```
